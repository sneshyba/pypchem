Orbitals such as these are the result of the application of a sophisticated algorithm to solve Schrödinger's equation (see the caption for the specific name). It turns out, however, that we can get a lot of insight into why these orbitals take on the shapes they do, using what's called the *Linear Combination of Atomic Orbitals* (or *LCAO*) method. The orbital shown in Fig. 1, for example, looks a lot like two $P_z$ orbitals, lined up with opposite orientation (phase) on the two oxygen atoms. 

## Computing $Z_{eff}$
To construct the orbitals used in LCAO, it turns out that we'll need to know $Z_{eff}$, the effective nuclear charge "felt" by a valence electron, for each atom in our molecule. Generally, the idea is that $Z_{eff}<Z_{true}$, because electrons prevent other electrons from experiencing the full nuclear charge of the nucleus. Two approaches to calculating $Z_{eff}$ can be identified -- of which the second of these is generally considered preferable.

- *Naive method*. Assume core electrons shield 100%, and other valence electrons shield nothing. In that case, the calculation is easy. In the case of carbon, for example, since $Z=6$ and there are two core electrons in the $n=1$ shell, then the effective nuclear charge felt by a valence ($n=2$) electron would be $Z=6-2=4$. 
- *Slater's method*. Core electrons don't shield 100% (which would make $Z_{eff}$ bigger than the naive prediction), but on the other hand, same-shell electrons shield more than zero (which would make $Z_{eff}$ smaller). Slater devised a formula estimating the net effect of these competing influences (see, e.g. https://chem.libretexts.org/Courses/Mount_Royal_University/Chem_1201/Unit_2._Periodic_Properties_of_the_Elements/2.06%3A_Slater's_Rules), for which there are online calculators (e.g., https://www.omnicalculator.com/chemistry/effective-charge).
