## Constructing orbital superpositions
*Why construct superpositions*  
Sympy's hydrogenic wave functions are eigenfunctions of hydrogenic hamiltonians (i.e. one-electron hamiltonians). That means they have "good quantum numbers" $n$, $l$, and $m$. But these functions are not always real-valued, and that can be awkward for visualising purposes. We can make real-valued orbitals from them by combining them as *linear superpositions*, but doing so comes at the cost of losing the $m$ quantum number. *C'est la vie*.

So how do we do that? To start with, Sympy's $\psi_{2,0,0}$ and $\psi_{2,1,0}$ are already real-valued, and equal to what chemists typically call $\psi_{2s}$ and $\psi_{2p_z}$. $\psi_{2,1,1}$ and $\psi_{2,1,-1}$ are complex-valued, however. In place of these, we can use the superpositions

$$
\psi_{2,1,1}-\psi_{2,1,-1} \equiv \psi_{2p_x}  \ \ \ \ (1)
$$

and 

$$
(\psi_{2,1,1}+\psi_{2,1,-1})/i  \equiv \psi_{2p_y} \ \ \ \ (2)
$$

where $i \equiv \sqrt{-1}$. In fact, we should be applying an additional normalization factor to each of these, so (for example) Eq. 1 should be replaced by ${1 \over N}(\psi_{2,1,1}-\psi_{2,1,-1})$, with $N=\sqrt2$. We're omitting the $N$ for simplicity, but in the work below you should remember to apply it. 

*About good quantum numbers*  
$\psi_{2s}$ and $\psi_{2p_z}$ both have three "good quantum numbers" ($n$, $l$, and $m$). And while $\psi_{2p_x}$ and $\psi_{2p_y}$ are made from wave functions with the same $n$ and $l$, hence those are still good quantum numbers, the $m$ values of the two contributors are different, and therefore $m$ is *not* a good quantum number for $\psi_{2p_x}$ and $\psi_{2p_y}$. What this means physically is that while $\psi_{2,1,1}$ and $\psi_{2,1,-1}$ imply an electron circling "to the right" or "to the left," we should understand $\psi_{2p_x}$ and $\psi_{2p_y}$ as implying some crazy superposition of those motions.

## Hybrid orbitals
You are probably also familiar with atomic orbital *hybrids*, which are superpositions involving orbitals of different angular moment quantum number. For example, the two distinct superpositions of an $s$ and a $p$ orbital are

$$
\psi_{sp_x(1)} =  \psi_{2ps} + \psi_{2p_y} \ \ \ \ (3)
$$
$$
\psi_{sp_x(2)} =  \psi_{2ps} - \psi_{2p_y} \ \ \ \ (4)
$$

We're saying "distinct* here because superpositions with overall sign change are not distinct (not *linearly independent* from each other -- an example would be $-\psi_{2ps} - \psi_{2p_y}$, which is not distinct from \psi_{sp_x(1)} as defined above. Another way of thinking about this is that there's a kind of conservation of wave functions: if what goes into a hybridization are two orbitals, then (only) two distinct hybrid orbitals can come out. 

Superpositions similar to those in Eqs. 3 and 4 yield $\psi_{sp_y(1)}$ and $\psi_{sp_y(2)}$. To make $sp^2$ hybrids, one set of superpositions is 

$$
\psi_{sp^2_xy(1)} = \psi_{2s} + 2^{1/2}\psi_{2p_y} \ \ \ \ (5)
$$
$$
\psi_{sp^2_xy(2)} = 2^{1/2}\psi_{2s} + 3^{1/2}\psi_{2p_x} - \psi_{2p_y} \ \ \ \ (6)
$$
$$
\psi_{sp^2_xy(3)} = 2^{1/2}\psi_{2s} - 3^{1/2}\psi_{2p_x} - \psi_{2p_y}  \ \ \ \ (7)
$$

(although that's not the only to construct $sp^2$ hybrids, as you'll see below).
